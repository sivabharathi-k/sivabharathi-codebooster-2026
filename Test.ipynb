```python
import pandas as pd
print("Libraries imported successfully!")
```

```python
df=pd.read_csv("student_performance.csv")
print(df)
```

```python
print("Total Student",df.shape[0])
print("Total Colums",df.shape[1])
print("Department",df['department'].unique())
```

```python
print(df["department"].value_counts())
```

```python
highest_score = df['math_score'].max()
print("Highest Score in Math",highest_score)

lowest_score = df['math_score'].min()
print("lowest Score in Math",lowest_score)

average_score = df['math_score'].mean().round(4)
print("Average Score in Math",average_score)

average_score = df['programming_score'].mean().round(2)
print("programming_score",average_score)





```

```python
print("\n[4] Gender-wise Average Scores")
print("-"*40)
gender_scores = df.groupby('gender')[['math_score', 'programming_score']].mean().round(2)
print(gender_scores)

```
